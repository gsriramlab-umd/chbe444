## 0. Introduction

This script plots the vapor-liquid equilibrium (VLE) field for the nonideal ternary system acetone/methanol/chloroform. 

Their Antoine parameters can be obtained from the NIST WebBook (https://webbook.nist.gov/chemistry). 

Because this is an ideal system, the activity coefficients $\gamma_i$ different from 1, and the Wilson equation parameters are generally different from zero. For a nonideal system, $\gamma_i$ can be obtained by the Wilson correlation:

$$
\ln{\gamma}_i = -\ln \left \{ \sum_{j=1}^{3} A_{ij} x_j \right \} + \sum_{j=1}^{3} \frac{A_{ji} x_j}{\sum_{k=1}^{3} A_{jk} x_k}
$$

where:

$$
\ln A_{ij} = a_{ij} + \frac{b_{ij}}{T}
$$

The parameters $a_{ij}$ and $b_{ij}$ are best obtained from Aspen. For this, create an Aspen compound (``apwz``) or worksheet (``apw``) file containing the three compounds of interest, go to ``Methods`` on the left menu pane and select ``WILSON`` as the method with default settings. Then navigate to ``Methods`` $\rightarrow$ ``Parameters`` $\rightarrow$ ``Binary Interaction`` $\rightarrow$ ``Wilson`` and read the parameters $a_{ij}$ and $b_{ij}$ from the displayed table. The parameters $a_{ij}$ and $b_{ij}$ should be set up as a matrix for non-ideal systems. Note that the diagonal elements  are always zero.

## 1. Specify Ternary System and Perform VLE Calculations

### 1.1 Specify ternary system

```python
import chbe444umd as des

%config InlineBackend.figure_format='svg'
%matplotlib inline

sysACM = des.TernarySystem(  # acetone/methanol/chloroform (non-ideal)
    component_ids={'acetone', 'methanol', 'chloroform'},
    component_names={'acetone', 'methanol', 'chloroform'},
    axes={'acetone': 0, 'methanol': 1},
    Antoine={'acetone':    {'A': 4.4245, 'B': 1312.3, 'C': -32.445},
             'methanol':   {'A': 5.2041, 'B': 1581.3, 'C': -33.5},
             'chloroform': {'A': 4.2077, 'B': 1233.1, 'C': -40.9530}},
    Wilson_a={'acetone': {'acetone':  0,
                          'methanol':  0,
                          'chloroform': -0.7683},
              'methanol': {'acetone': 0,
                           'methanol':  0,
                           'chloroform':  0},
              'chloroform': {'acetone': -0.7191,
                             'methanol':  0,
                             'chloroform':  0}},
    Wilson_b={'acetone':    {'acetone':  0,
                             'methanol': -115.663,
                             'chloroform':  262.1790},
              'methanol':   {'acetone': -108.5260,
                             'methanol':  0,
                             'chloroform': -652.8960},
              'chloroform': {'acetone':  435.1440,
                             'methanol': -32.5972,
                             'chloroform':  0}}
)
```

### 1.2 Plot equilibrium field

```python
fig, ax, _ = sysACM.simulate_vle_field(
    n_vectors=51, arrow_scale=51,
    Tmin=50, Tmax=70,
    contour_T_heavy=1, contour_T_medium=0.5, contour_T_light=0.1)

# plt.savefig('ACM.svg', format='svg', bbox_inches='tight', pad_inches=0)
```

### 1.3 Query VLE properties

```python
sysACM.gamma([[0.3, 0.4, 0.3], [0.2, 0.1, 0.7]], [40, 55])
```

```python
print(sysACM.vle_properties([0.32, 0.4, 0.28]))
```

## 2. Integrate Residue Curves

```python
ax, res_fwd, res_bwd = sysACM.integrate_residue(
    ax, feed={
        0: {'acetone': 0.4, 'methanol': 0.2, 'chloroform': 0.4},
        1: {'acetone': 0.6, 'methanol': 0.2, 'chloroform': 0.2},
        2: {'acetone': 0.2, 'methanol': 0.6, 'chloroform': 0.2},
        3: {'acetone': 0.5, 'methanol': 0.4, 'chloroform': 0.1}},
    integration_time=20, max_step=0.5, show_feed=True,
    linestyle='--', color='k', marker='o', markersize=6)
fig
```

## 3. Analyze Azeotropy

### 3.1 Find stationary points

```python
res = sysACM.find_stationary_points(decimals=2)
print(res)
```

```python
for x in res.x:
    ax.plot(x[0], x[1], 'go')
fig
```

### 3.2 Plot separatrices

```python
sysACM.plot_separatrices(ax, stationary_points=res, c='g', ls='-')
fig
```
