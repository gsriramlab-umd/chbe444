## Simplex Method

This function from the course package `chbe444umd` simulates the simplex algorithm to solve a given LP.

```python
# This package uses the command ``simplex_tableaux`` from ``chbe444umd``
# To find out about this command and its syntax, type the command followed by ?

import chbe444umd as des  # import course package
des.simplex_tableaux?
```

```python
# Solution of the urea reaction path LP by simplex algorithm
c = [0, -1 / 2]
A = [[1, -1 / 2],
     [-1 / 3, 1 / 2],
     [-1, 0]]
b = [0, 1, -1]
var_names = ['x1', 'x3']  # custom names for original variables
slack_names = ['x4', 'x5', 'x6']  # custom names for slack variables

simplex = des.simplex_tableaux(c, A_ub=A, b_ub=b,
                               var_names=var_names,
                               slack_names=slack_names,
                               tol=1e-6, dec=3,
                               show_tableaux=True, colormap='YlGn')

simplex  # print the object generated by the algorithm
```
