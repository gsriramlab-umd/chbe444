```python
import numpy as np
import scipy.optimize as opt


def z(x):
    z = 5 * (x[0] - 4)**2 + 10 * (x[1] - 2)**2
    return z


# z = lambda x: (5 * (x[0] - 4)**2 + 10 * (x[1] - 2)**2)  # lambda format

z([1, 1])  # test the function

ineq_cons = {'type': 'ineq',
             'fun': lambda x: -np.array([-x[0],
                                         -x[1],
                                         x[0] - 2,
                                         x[1] - 4,
                                         3 * x[0]**2 + 2*x[1] - 6,
                                         -x[0] - x[1] + 2])}

x0 = [0, 1]

res = opt.minimize(z, x0, method='SLSQP',
                   constraints=[ineq_cons],
                   options={'ftol': 1e-9, 'disp': True})

print(res.x)  # location (coordinates) of the optimum
print(res.fun)  # optimal value of z
print(res.success)  # True/False value
```
